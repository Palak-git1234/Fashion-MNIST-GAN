```python
# ============================================================
# GAN: How the Generator and Discriminator Work (Fashion-MNIST)
# ============================================================
# Generator (G)     : random noise -> fake image. Goal: fool D.
# Discriminator (D) : image -> real or fake score. Goal: catch G.

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers
from tensorflow.keras.datasets import fashion_mnist

# ---------------- 1. Load and prepare real data ----------------
(x_train, _), _ = fashion_mnist.load_data()
x_train = (x_train.astype("float32") - 127.5) / 127.5   # [0,255] -> [-1,1] (matches tanh)
x_train = x_train.reshape(-1, 28, 28, 1)                # add channel dimension

BATCH_SIZE = 128
NOISE_DIM = 100                                         # size of random noise for G
train_dataset = (tf.data.Dataset.from_tensor_slices(x_train)
                 .shuffle(60000)
                 .batch(BATCH_SIZE, drop_remainder=True))

def show_images(images, title=""):
    plt.figure(figsize=(6, 6))
    for i in range(16):
        plt.subplot(4, 4, i + 1)
        plt.imshow((images[i, :, :, 0] + 1) / 2, cmap="gray")
        plt.axis("off")
    plt.suptitle(title)
    plt.show()

show_images(x_train, "REAL images from the dataset")

# ---------------- 2. Generator ----------------
# Starts small and GROWS the image: 100 numbers -> 7x7 -> 14x14 -> 28x28
def build_generator():
    model = tf.keras.Sequential(name="Generator")
    model.add(layers.Input(shape=(NOISE_DIM,)))
    model.add(layers.Dense(7 * 7 * 128, use_bias=False))      # expand the noise
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU())
    model.add(layers.Reshape((7, 7, 128)))                    # tiny 7x7 "image"
    model.add(layers.Conv2DTranspose(64, 5, strides=2, padding="same", use_bias=False))  # 7 -> 14
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU())
    model.add(layers.Conv2DTranspose(1, 5, strides=2, padding="same",
                                     use_bias=False, activation="tanh"))                 # 14 -> 28
    return model

generator = build_generator()
generator.summary()

# Untrained generator = random weights = just static
noise = tf.random.normal([16, NOISE_DIM])
untrained_fake = generator(noise, training=False)
show_images(untrained_fake.numpy(), "Untrained GENERATOR output (just noise)")

# ---------------- 3. Discriminator ----------------
# Opposite of G: SHRINKS the image: 28x28 -> 14x14 -> 7x7 -> one score
def build_discriminator():
    model = tf.keras.Sequential(name="Discriminator")
    model.add(layers.Input(shape=(28, 28, 1)))
    model.add(layers.Conv2D(64, 5, strides=2, padding="same"))    # 28 -> 14
    model.add(layers.LeakyReLU())
    model.add(layers.Dropout(0.3))                                # stops D getting too strong too fast
    model.add(layers.Conv2D(128, 5, strides=2, padding="same"))   # 14 -> 7
    model.add(layers.LeakyReLU())
    model.add(layers.Dropout(0.3))
    model.add(layers.Flatten())
    model.add(layers.Dense(1))                                    # raw score (positive = real)
    return model

discriminator = build_discriminator()
discriminator.summary()

# Untrained discriminator is just guessing (about 0.5)
real_score = tf.sigmoid(discriminator(x_train[:16], training=False))
fake_score = tf.sigmoid(discriminator(untrained_fake, training=False))
print("D says REAL images are real with probability:", round(float(tf.reduce_mean(real_score)), 2))
print("D says FAKE images are real with probability:", round(float(tf.reduce_mean(fake_score)), 2))

# ---------------- 4. Losses and training step ----------------
# D wants: real -> 1, fake -> 0
# G wants: its fake -> 1 (fool D)
cross_entropy = tf.keras.losses.BinaryCrossentropy(from_logits=True)

def discriminator_loss(real_output, fake_output):
    real_loss = cross_entropy(tf.ones_like(real_output), real_output)
    fake_loss = cross_entropy(tf.zeros_like(fake_output), fake_output)
    return real_loss + fake_loss

def generator_loss(fake_output):
    return cross_entropy(tf.ones_like(fake_output), fake_output)

g_opt = tf.keras.optimizers.Adam(0.0002, beta_1=0.5)
d_opt = tf.keras.optimizers.Adam(0.0002, beta_1=0.5)

@tf.function
def train_step(real_images):
    noise = tf.random.normal([tf.shape(real_images)[0], NOISE_DIM])

    with tf.GradientTape() as g_tape, tf.GradientTape() as d_tape:
        fake_images = generator(noise, training=True)           # 1. G makes fakes
        real_out = discriminator(real_images, training=True)    # 2. D scores real images
        fake_out = discriminator(fake_images, training=True)    #    D scores fake images
        g_loss = generator_loss(fake_out)                       # 3. calculate both losses
        d_loss = discriminator_loss(real_out, fake_out)

    g_grads = g_tape.gradient(g_loss, generator.trainable_variables)
    d_grads = d_tape.gradient(d_loss, discriminator.trainable_variables)
    g_opt.apply_gradients(zip(g_grads, generator.trainable_variables))      # 4. update both
    d_opt.apply_gradients(zip(d_grads, discriminator.trainable_variables))

    return (g_loss, d_loss,
            tf.reduce_mean(tf.sigmoid(real_out)),    # D's confidence on real images
            tf.reduce_mean(tf.sigmoid(fake_out)))    # D's confidence on fake images

# ---------------- 5. Training ----------------
EPOCHS = 20
SHOW_AT = [0, 1, 5, 10, 20]

seed = tf.random.normal([16, NOISE_DIM])      # same noise every epoch to track progress
snapshots = {0: generator(seed, training=False).numpy()}
d_real_hist, d_fake_hist = [], []

for epoch in range(1, EPOCHS + 1):
    r, f = [], []
    for batch in train_dataset:
        g_loss, d_loss, d_real, d_fake = train_step(batch)
        r.append(d_real.numpy())
        f.append(d_fake.numpy())

    d_real_hist.append(np.mean(r))
    d_fake_hist.append(np.mean(f))
    print(f"Epoch {epoch:2d}/{EPOCHS} | G loss {g_loss:.2f} | D loss {d_loss:.2f} | "
          f"D(real) {d_real_hist[-1]:.2f} | D(fake) {d_fake_hist[-1]:.2f}")

    if epoch in SHOW_AT:
        snapshots[epoch] = generator(seed, training=False).numpy()

# ---------------- 6a. Result: generator improves ----------------
fig, axes = plt.subplots(1, len(SHOW_AT), figsize=(3.2 * len(SHOW_AT), 3.6))
for ax, e in zip(axes, SHOW_AT):
    imgs = (snapshots[e][:, :, :, 0] + 1) / 2
    grid = np.vstack([np.hstack(imgs[r * 4:(r + 1) * 4]) for r in range(4)])
    ax.imshow(grid, cmap="gray")
    ax.set_title("Before training" if e == 0 else f"Epoch {e}")
    ax.axis("off")
plt.suptitle("Generator progress (same noise every time)")
plt.tight_layout()
plt.show()

# ---------------- 6b. Result: discriminator gets fooled ----------------
plt.plot(d_real_hist, label="D(real images)")
plt.plot(d_fake_hist, label="D(fake images)")
plt.axhline(0.5, color="gray", linestyle="--", label="0.5 = D is just guessing")
plt.xlabel("Epoch")
plt.ylabel("Probability of 'real'")
plt.title("Discriminator confidence")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# ---------------- 6c. Result: real vs fake ----------------
real_samples = x_train[np.random.choice(len(x_train), 10, replace=False)]
fake_samples = generator(tf.random.normal([10, NOISE_DIM]), training=False).numpy()

plt.figure(figsize=(15, 3.5))
for i in range(10):
    plt.subplot(2, 10, i + 1)
    plt.imshow((real_samples[i, :, :, 0] + 1) / 2, cmap="gray")
    plt.axis("off")
    plt.subplot(2, 10, 10 + i + 1)
    plt.imshow((fake_samples[i, :, :, 0] + 1) / 2, cmap="gray")
    plt.axis("off")
plt.suptitle("Top row: REAL   |   Bottom row: FAKE (made by generator)")
plt.tight_layout()
plt.show() 
```
